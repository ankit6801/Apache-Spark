Answer the following questions using sf_fire_calls table

   Create Dataframe from the table to start answering

```python
fire_calls_df = spark.read.table("dev_catalog.spark_db.sf_fire_calls")
```

Q1. How many distinct types of calls were made to the Fire Department?

```python
%sql
select count(distinct CallType) as dictinct_call_Type from dev_catalog.spark_db.sf_fire_calls
```

```python
q1_df = (
    fire_calls_df.selectExpr("count(distinct CallType) as distinct_callType")
)
q1_df.show()
```

Q2. What were distinct types of calls made to the Fire Deartment

```python
%sql
select distinct CallType as distinct_callType from dev_catalog.spark_db.sf_fire_calls
where CallType is not null
```

```python
q2_df = (
fire_calls_df.selectExpr("CallType as distinct_callType")
.distinct()
)
q2_df.display()
```

3. Find out all response for delayed times greater than 5 mins?

```python
%sql
select CallNumber,Delay from dev_catalog.spark_db.sf_fire_calls
where Delay >5
```

```python
(fire_calls_df.where("Delay > 5")
.select("CallType", "Delay")
.display()
)
```

Q4. What were the most common call types?

```python
%sql
select CallType, count(*) as count from dev_catalog.spark_db.sf_fire_calls
where CallType is not null 
group by  CallType
order by count desc
    
```

```python
(
    fire_calls_df.select("CallType")
    .where ("CallType is not null")
    .groupBy("CallType").count()
    .orderBy("count", ascending=False)
    .display()
)

```

Q5. What zip codes accounted for most common calls?

```python
%sql
select CallType,zipcode,count(*) as count from dev_catalog.spark_db.sf_fire_calls
where CallType is not null
group by Zipcode,CallType
order by count desc
```

```python
(
    fire_calls_df.select("CallType","Zipcode")
    .where ("CallType is not null")
    .groupBy("CallType","Zipcode").count()
    .orderBy("count", ascending=False)
    .display()

)
```

Q6. What San Francisco neighbourhoods are in the zip codes 94102 and 94103.

```python
%sql
select distinct Neighborhood,Zipcode from dev_catalog.spark_db.sf_fire_calls
where Zipcode == '94102' or  Zipcode == '94103'
```

```python
(
    fire_calls_df.select("Neighborhood","Zipcode")
    .where ("zipcode == 94102 or zipcode = 94103" )
    .distinct()
    .display()
)
```

Q8. How many distinct years of data is in csv file?

```python
%sql
select distinct year(CallDate) as year_num from dev_catalog.spark_db.sf_fire_calls
order by year_num
```

```python
(
    fire_calls_df.selectExpr("year(CallDate) as year_num")
    .distinct()
    .orderBy("year_num")
    .show()
)
```

Q9. What week of the year in 2018 had the most fire calls?

```python
%sql
select weekofyear(CallDate) as week_year, count(*) as count
from dev_catalog.spark_db.sf_fire_calls
where year(CallDate) == 2018
group by week_year
order by count desc
```

```python
(
    fire_calls_df.selectExpr("weekofyear(CallDate) as Week_year")
    .where ("year(CallDate) == 2018")
    .groupBy("Week_year")
    .count()
    .orderBy("count",ascending=False)
    .display()
)
```

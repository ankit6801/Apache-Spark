Requirement:

Answer the below question using the dev_catalog.spark_db.sf_fire_calls
What are top 3 zip code that accounted for most calls?

```python
%sql
select CallType,zipcode ,count(*) as count from dev_catalog.spark_db.sf_fire_calls
where CallType is not null
group by Zipcode,CallType
order by count desc
limit 3
```

2. Answer the above question using the Spark Datafram API

2.1 Create a Dataframe reading data from the table.

```python
# Read data and create a dataframe
fire_data_df = spark.read.table("dev_catalog.spark_db.sf_fire_calls")
```

2.2 Apply Transformations

```python
# Apply necessary tranformations
df_1 = fire_data_df.select("CallType","Zipcode")
df_2 = df_1.where("CallType is not NULL")
df_3 = df_2.groupBy("CallType","Zipcode").count()
df_4 = df_3.orderBy("total_count",ascending=False)
df_5 = df_4.limit(3)
df_5.display()
```

3. Spark Dataframe API concepts
  
     1. Spark creates optimized query plan
     2. Dataframe are immutable
     3. Every Transformation return a dataframe
     4. Dataframe offers composable API .

Spark creates optimized query plan
→ Spark analyzes all transformations and generates an optimized execution plan using Catalyst Optimizer before running the job.

DataFrames are immutable
→ You cannot modify a DataFrame in place; any change creates a new DataFrame.

Every transformation returns a DataFrame
→ Operations like select, filter, withColumn don’t change data directly; they return a new DataFrame.

DataFrame offers composable API
→ You can chain multiple operations together (method chaining) to build complex logic step by step.

One line summary:
Spark DataFrames are immutable, lazily optimized datasets where transformations compose into an efficient execution plan.

5. Example of composable API

```python
result_data_df = (
    fire_data_df.select("CallType","Zipcode")
    .where("CallType is not null")
    .groupBy("CallType","Zipcode").count()
    .orderBy("count",ascending=False)
    .limit(3)
)
```

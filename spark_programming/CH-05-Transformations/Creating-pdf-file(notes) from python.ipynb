```python
# Retry generating the Hinglish revision notes PDF

from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.pagesizes import A4



file_path = "/Workspace/Users/ankitgre1904@gmail.com/Apache-Spark/Notes(Key Point)/PySpark_Databricks_Hinglish_Revision_Notes.pdf"

doc = SimpleDocTemplate(
    file_path,
    pagesize=A4,
    rightMargin=36,
    leftMargin=36,
    topMargin=36,
    bottomMargin=36,
)

styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name="TitleStyle", fontSize=18, leading=22, spaceAfter=14))
styles.add(ParagraphStyle(name="HeaderStyle", fontSize=13, leading=16, spaceBefore=12, spaceAfter=6))
styles.add(ParagraphStyle(name="BodyStyle", fontSize=10.5, leading=14, spaceAfter=6))

content = []

content.append(Paragraph("PySpark & Databricks – Hinglish Revision Notes", styles["TitleStyle"]))

notes = [
("1. DataFrame & Lazy Execution",
"DataFrame ek distributed table hota hai. Spark tab tak kuch run nahi karta jab tak action call na ho. Isko lazy execution kehte hain. Isse performance better hoti hai."),

("2. Transformation vs Action",
"Transformation sirf logical plan banata hai (select, filter, withColumn). Action job ko actually run karta hai (collect, count, show, display)."),

("3. select() vs selectExpr()",
"select() PySpark Column API hai – safe. selectExpr() SQL style hai. Same select/selectExpr me alias reuse nahi hota (alias dependency rule)."),

("4. withColumn()",
"withColumn naya column bana sakta hai ya existing column overwrite kar sakta hai. Chained withColumn sequential hota hai – alias dependency issue nahi aata."),

("5. withColumnRenamed()",
"Sirf column ka naam change karta hai. Koi bhi column drop nahi hota. Rename ke liye best choice."),

("6. Alias Dependency",
"Same select/selectExpr me alias reuse karna galat hai. Expression repeat karo ya step-by-step withColumn use karo."),

("7. cast() vs try_cast()",
"cast strict hota hai aur error deta hai. try_cast safe hai aur NULL return karta hai. Production me try_cast prefer karo."),

("8. coalesce()",
"Pehla non-null value return karta hai. Default value dene aur NULL handle karne ke liye use hota hai."),

("9. expr()",
"Spark SQL expressions likhne ke liye use hota hai, jaise case when, try_cast."),

("10. case when",
"SQL ka IF-ELSE logic. Hamesha END se close hota hai."),

("11. Date Parsing",
"Spark default date format yyyy-MM-dd hai. Alag format ke liye to_date(col, format) use karo."),

("12. distinct() vs dropDuplicates()",
"distinct poori row ke basis pe duplicates hatata hai. dropDuplicates(['id']) column-based duplicate remove karta hai."),

("13. StructType Schema",
"Proper schema definition deta hai with datatype & nullable control. Production me best practice."),

("14. Data Types",
"Primitive: StringType, IntegerType, LongType, DoubleType, DateType. Complex: ArrayType, MapType, StructType."),

("15. JSON vs TEXT read",
"json format data parse karta hai aur columns banata hai. text sirf raw string deta hai."),

("16. uuid4() + lit()",
"uuid4 random unique id deta hai. lit Python value ko Spark column banata hai. Batch_id ke liye use hota hai."),

("17. collect()",
"Action hai jo data driver pe laata hai. Sirf small data ke liye safe."),

("18. Row & asDict()",
"collect ke baad har row ek Row object hota hai. asDict() row ko Python dictionary me convert karta hai."),

("19. Filtering rules",
"PySpark me & | ~ operators use karo. and/or Python ke hote hain – avoid karo."),

("20. Golden Rules",
"Alias reuse avoid karo, try_cast prefer karo, collect limited use karo, StructType use karo.")
]

for h, b in notes:
    content.append(Paragraph(h, styles["HeaderStyle"]))
    content.append(Paragraph(b, styles["BodyStyle"]))

doc.build(content)

file_path
```

```python
from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet

file_path = "/Workspace/Users/ankitgre1904@gmail.com/Apache-Spark/Notes(Key Point)/pyspark_notes_v1.pdf"
doc = SimpleDocTemplate(file_path)
styles = getSampleStyleSheet()
content = []

summary_text = """
PySpark & Databricks Short Notes

1. cast() vs try_cast()
cast = strict, try_cast = safe (returns NULL)

2. coalesce()
Returns first non-null value

3. expr()
Used to write Spark SQL expressions

4. select vs selectExpr
select = PySpark style
selectExpr = SQL style

5. Alias Dependency
Do not reuse aliases in same select/selectExpr

6. withColumnRenamed
Only renames column, does not drop others

7. StructType
Defines schema with full control

8. Data Types
StringType, IntegerType, DoubleType, DateType, ArrayType, MapType, StructType

9. case when
SQL conditional logic, always ends with END

10. uuid4 + lit
Used to generate batch_id

Golden Rule:
Prefer safe, step-by-step transformations in production.
"""

content.append(
    Paragraph(summary_text.replace("\n", "<br/>"), styles["Normal"])
)

doc.build(content)

file_path
```

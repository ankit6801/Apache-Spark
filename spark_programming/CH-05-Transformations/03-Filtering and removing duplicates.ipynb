````python
from pyspark.sql.types  import*
data_schema = StructType([
    StructField("id", IntegerType(), False),
    StructField("source", StringType(), True),
    StructField("destination", StringType(), True),
    StructField("distance", IntegerType(), True)
])

data_list = [
    [101, "Chicago", "New York", 700],
    [102, "Chicago", "Los Angeles", 2000],
    [102, "Chicago", "Los Angeles", 2000],
    [103, "Mumbai", "Chennai", 1200],
    [104, "Mumbai", "Chennai", 1200],
    [105, "Paris", "Phoenix", 1800],
    [105, "Paris", "Phoenix", 1800]
]
df = spark.createDataFrame(data_list, data_schema)
display(df)
````

## 2. Select records that satisfy the following criteria
1. Source is Chicago and destination is Los Angeles

Show the following approaches
1. Using a single filter
2. Using two filters
3. Using a col function expression
4. Using a dataframe qualifier expression

2.1 Using a single filter.

````python
df.filter("source = 'Chicago' and destination = 'Los Angeles'").display()
````

2.2 Using two filters

````python
df.filter("source = 'Chicago'").filter("destination = 'Los Angeles'").display()
````

2.3 Using a col function expression

````python
from pyspark.sql.functions import*
df.filter( (col("source")=="Chicago")  &  (col("destination")=="Los Angeles") ).display()
````

2.4 Using a dataframe qualifier expressionn

````python
df.filter((df.source == "Chicago") & (df.destination == "Los Angeles")).display()
````

## One-line interview answer
Spark provides numeric, string, date/time, boolean, and complex data types like ArrayType, MapType, and StructType to define structured schemas.

## Easy memory trick

Number → Integer / Long / Double
Text → StringType
Date → DateType / TimestampType
List → ArrayType
Key-Value → MapType
Nested row → StructType



## 1. Numeric Types
```
| DataType            | Meaning                  |
| ------------------- | ------------------------ |
| `ByteType()`        | 1-byte integer           |
| `ShortType()`       | 2-byte integer           |
| `IntegerType()`     | normal integer           |
| `LongType()`        | large integer            |
| `FloatType()`       | decimal (less precision) |
| `DoubleType()`      | decimal (more precision) |
| `DecimalType(p, s)` | fixed precision decimal  |
```
## 2. String/Binary Type
```
| DataType       | Meaning             |
| -------------- | ------------------- |
| `StringType()` | text / string       |
| `BinaryType()` | binary data (bytes) |
```
## 3. Boolean Type
```
| DataType        | Meaning      |
| --------------- | ------------ |
| `BooleanType()` | True / False |
```
## 4. Date & Time Types
```
| DataType             | Meaning                    |
| -------------------- | -------------------------- |
| `DateType()`         | date only                  |
| `TimestampType()`    | date + time                |
| `TimestampNTZType()` | timestamp without timezone |
```
## 5. Collection / Complex Types
```
ArrayType(StringType())
```
## 6. Null /Special Types
```
| DataType                 | Meaning                    |
| ------------------------ | -------------------------- |
| `NullType()`             | only null values           |
| `CalendarIntervalType()` | interval (days, hours etc) |
```
## 7. User Define Type
```
| DataType            | Meaning     |
| ------------------- | ----------- |
| `UserDefinedType()` | custom type |
```

`````python
from pyspark.sql.types import StructField,StructType,IntegerType,ShortType,DoubleType, StringType
file_schema = "id int,name string,age short,salary double"

schema = StructType([
    StructField("id",IntegerType()),
    StructField("name",StringType()),
    StructField("age",ShortType()),
    StructField("salary",DoubleType())
])

data_list = [(101,'Alice',20,100000),
             (102,'Bob',30,200000),
             (103,'Charlie',40,300000 )]

# sample_df = spark.createDataFrame(data_list,["id","name","age","salary"])
# sample_df = spark.createDataFrame(data_list,file_schema)
sample_df = spark.createDataFrame(data_list,schema)
sample_df.show()

`````

2. Add following columns to your Dataframe

   1. increment: 10% of the salary up to 3000 maximum increment
   2. revised_salary: salary + increment

`````python
from pyspark.sql.functions import expr

salary_df = (
    sample_df.withColumns({
        "increment" : expr("case when salary > 30000 then 3000 else salary*10/100 end"),
        "revised_salary" : expr("salary+increment")
    })
)

salary_df.display()
`````

3. Add following columns to your Dataframe
     1. increment: 10% of the salary up to 3000 maximum increment
     2. Replace the following column in your dataframe
          1. alary: current salary + increment

`````python
from pyspark.sql.functions import expr

salary_df = (sample_df
.withColumn("increment", expr("case when salary > 30000 then 3000 else salary*10/100 end"))
.withColumn("salary",expr("salary + increment"))
)

salary_df.display()
`````

4. Add a batch number (uuid) column to your dataframe

`````python
import uuid
from pyspark.sql.functions import lit
batch_id = str(uuid.uuid4())

salary_batch_df = (sample_df
.withColumn("batch_id",lit(batch_id))
)
salary_batch_df.display()
`````

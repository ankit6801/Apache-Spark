1. Setup Catalog, Database and Volume
   1. Create a catalog - dev_catalog
   2. Create a database - dev_catalog.spark_db
   3. Create a volume - dev_catalog.spark_db.datasets

```python
%sql
create catalog if not exists dev;
create catalog if not exists dev_catalog;
create database if not exists dev_catalog.spark_db;
create volume if not exists dev_catalog.spark_db.datasets;
```

2. Import data ingestion utility

```python
%run ../utils/data-ingestion
```

3. Ingest Course Datasets

```python
DL = DataLoader()
DL.clean_ingest_data("spark_programming/data","spark_programming/data")
```

4. Create a table


```python
%sql
create table dev_catalog.spark_db.diamonds(
  carat double,
  clarity string,
  color string,
  cut string,
  depth string,
  price double
)
```

5. Load course data

```python
%sql
insert into dev_catalog.spark_db.diamonds
select * from json.`/Volumes/dev_catalog/spark_db/datasets/spark_programming/data/diamonds.json`
```

6. Query the table


```python
%sql
select color, avg(price) as avg_price from dev_catalog.spark_db.diamonds group by color order by avg_price desc
-- limit 3
```

####Join types in Apache Spark
1. Inner Join - Return only the rows that have matching values on both sides
2. Outer Joins
    * Left - Returns all rows from the left side and matching rows from the right side
    * Right - Returns all rows from the right side and matching rows from the left side
    * Full - Return all matching and non-matching rows from both sides
3. Natural Join - Automatically create join criteria on the same column names (Applies to Inner and Outer Joins)
4. Cross Join - Join without any join criteria (all possible combinations)
5. Self Join - Join a table with itself (Applies to Inner, Outer, and Cross Joins)
6. Semi Join - Take records from the left side when it matches with the right side
7. Anti Join - Take records from the left side when it doesn not match with the right side
8. Lateral Join - Each row from the driving table is used to subquery the derived table

# 2. Define schema

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType, DateType, TimestampType
facilities_schema = StructType([
    StructField('facility_id',IntegerType()),
    StructField('facility_name',StringType()),
    StructField('member_cost',DoubleType()),
    StructField('guest_cost',DoubleType()),
    StructField('initial_outlay',DoubleType()),
    StructField('monthly_maintainance',DoubleType()),

])

members_schema = StructType([
    StructField('member_id',IntegerType()),
    StructField('last_name',StringType()),
    StructField('first_name',StringType()),
    StructField('Address',StringType()),
    StructField('zip_code',StringType()),
    StructField('telephone',StringType()),
    StructField('recommended_by',StringType()),
    StructField('joining_date',DateType())
])

bookings_schema = StructType([
    StructField('booking_id',IntegerType()),
    StructField('facility_id',IntegerType()),
    StructField('member_id',IntegerType()),
    StructField('start_time',TimestampType()),
    StructField('slots',IntegerType())
    
])
```

# 3. Load facilities table

```python
facilities_df = (spark.read.format("csv")
.option("header","true")
.schema(facilities_schema)
.load("/Volumes/dev_catalog/spark_db/datasets/spark_programming/data/facilities.csv"))

# display(facilities_df)

facilities_df.write.mode("overwrite").saveAsTable("dev_catalog.spark_db.facilities")
```

4. Load members table

```python
members_df = (spark.read.format('csv')
.option("header","true")
.schema(members_schema)
.load('/Volumes/dev_catalog/spark_db/datasets/spark_programming/data/members.csv')
.withColumnRenamed('slot','Slots')
)

# display(members_df)
members_df.write.mode("overwrite").saveAsTable("dev_catalog.spark_db.members")
```

5. Load booking table



```python
bookings_df = (
    spark.read.format('csv')
    .option("header","true")
    .schema(bookings_schema)
    .load('/Volumes/dev_catalog/spark_db/datasets/spark_programming/data/bookings.csv')
    .withColumnRenamed('slot','Slots')
)

bookings_df.write.mode('overwrite').saveAsTable("dev_catalog.spark_db.bookings")
bookings_df = bookings_df.drop("slot")
display(bookings_df)
```
